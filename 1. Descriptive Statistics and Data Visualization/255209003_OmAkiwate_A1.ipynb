```python
def average(values):
    total =0
    for value in values:
        total+=value
    avg=total/len(values)
    
    return avg

print("Average of Test 1 : ", average(test1))
print("Average of Test 2 : ", average(test2))
print("Average of Attendance : ", average(attendance))
```

```python
import numpy as np
t1= np.array(test1)
t2= np.array(test2)
t3=np.array(attendance)
print(t2-t1)
```

```python
print("Mean : " , np.mean(t1))
print("Variance : ", np.var(t1))
print("STD Deviation : ", np.std(t1))
```

```python
students = np.column_stack((t1, t2, t3))

nisha = students[-1]   

min_distance = float('inf')
closest_student = ""

for i in range(len(names) - 1): 
    distance = np.sqrt(np.sum((students[i] - nisha) ** 2))

    if distance < min_distance:
        min_distance = distance
        closest_student = names[i]

print("Student closest to Nisha:", closest_student)
print("Distance:", min_distance)
```
